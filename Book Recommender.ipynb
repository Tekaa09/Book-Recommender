```python
# ==========================================================================
# HÀM XỬ LÝ DNA TÁC GIẢ
# ==========================================================================
import re
from collections import Counter
from datetime import datetime

NAM_NAY = datetime.now().year

# Nhóm tác giả cho bộ lọc trong trang DNA: khoá -> (Tiếng Việt, English)
NHOM_DNA = {
    "tat_ca":       ("Tất cả", "All"),
    "vn":           ("Việt Nam", "Vietnam"),
    "the_gioi":     ("Văn học thế giới", "World literature"),
    "truyen_tranh": ("Truyện tranh", "Comics & manga"),
    "ky_nang":      ("Triết học & kỹ năng", "Philosophy & self-help"),
}

# Chữ hiển thị trên giao diện DNA (nội dung tiểu sử giữ tiếng Việt)
CHU_DNA = {
    "vi": {
        "ho_so": "HỒ SƠ DNA", "su_kien": "SỰ KIỆN", "tac_pham": "TÁC PHẨM",
        "sinh_nhan": "KHỞI ĐẦU", "mat_nhan": "KHÉP LẠI", "sinh": "Chào đời", "mat": "Qua đời", "tai": "Tại",
        "tuoi": "TUỔI ĐỜI", "tuoi_nay": "TUỔI HIỆN TẠI", "so_tp": "TÁC PHẨM TRONG THƯ VIỆN",
        "so_moc": "MỐC THỜI GIAN", "gen_troi": "GEN TRỘI",
        "gen": "BẢN ĐỒ GEN SÁNG TÁC", "gen_phu": "Tỉ lệ tác phẩm trong thư viện mang mỗi chủ đề",
        "chua_co_sach": "Thư viện chưa có tác phẩm nào của tác giả này — hãy bổ sung vào books_300.json.",
        "cung_gen": "TÁC GIẢ CÙNG “GEN”", "cung_gen_phu": "Bấm để mở hồ sơ · xếp theo độ trùng chủ đề",
        "khong_lq": "Chưa tìm thấy tác giả có chủ đề tương đồng.",
        "hanh_trinh": "HÀNH TRÌNH", "tat_ca": "Tất cả", "loc_sk": "Sự kiện đời", "loc_tp": "Tác phẩm",
        "trang": "trang", "trong_tv": "Có trong thư viện",
        "chua_co": "Chưa có dữ liệu DNA cho tác giả này.", "nay": "nay", "tcn": "TCN",
        "toan_canh": "TOÀN CẢNH CUỘC ĐỜI",
        "nguon": "Dữ liệu DNA do nhóm biên soạn · Ảnh chân dung lấy từ Wikipedia (nếu có)",
        "banner_kicker": "GIẢI MÃ HÀNH TRÌNH SÁNG TÁC", "banner_tieu_de": "DNA TÁC GIẢ",
        "banner_mo_ta": "{n} hồ sơ tác giả · {s} cuốn sách được nối vào từng mốc đời — "
                        "mỗi sự kiện là một “gen”, mỗi tác phẩm là một đoạn mã.",
    },
    "en": {
        "ho_so": "DNA PROFILE", "su_kien": "EVENT", "tac_pham": "WORK",
        "sinh_nhan": "BEGINNING", "mat_nhan": "FAREWELL", "sinh": "Born", "mat": "Passed away", "tai": "In",
        "tuoi": "LIFESPAN", "tuoi_nay": "CURRENT AGE", "so_tp": "WORKS IN LIBRARY",
        "so_moc": "MILESTONES", "gen_troi": "DOMINANT GENE",
        "gen": "CREATIVE GENE MAP", "gen_phu": "Share of this author's books tagged with each topic",
        "chua_co_sach": "The library has no books by this author yet — add some to books_300.json.",
        "cung_gen": "AUTHORS WITH SIMILAR “GENES”", "cung_gen_phu": "Click to open · ranked by topic overlap",
        "khong_lq": "No authors with similar topics found yet.",
        "hanh_trinh": "JOURNEY", "tat_ca": "All", "loc_sk": "Life events", "loc_tp": "Works",
        "trang": "pages", "trong_tv": "In the library",
        "chua_co": "No DNA data for this author yet.", "nay": "now", "tcn": "BC",
        "toan_canh": "LIFE AT A GLANCE",
        "nguon": "DNA data compiled by the team · Portraits from Wikipedia (when available)",
        "banner_kicker": "DECODING CREATIVE JOURNEYS", "banner_tieu_de": "AUTHOR DNA",
        "banner_mo_ta": "{n} author profiles · {s} books linked to each life milestone — "
                        "every event is a “gene”, every work a strand of code.",
    },
}

# Bảng màu cho các thanh "gen" (biến màu định nghĩa trong CSS mục 11)
MAU_GEN = ["var(--d-teal)", "var(--d-amber)", "var(--d-tim)", "var(--d-hong)", "var(--d-xanh)", "var(--d-la)"]
MAU_MOC = {"tac_pham": "var(--d-teal)", "su_kien": "var(--d-amber)", "sinh": "var(--d-la)", "mat": "var(--d-tim)"}
THU_TU_MOC = {"sinh": 0, "su_kien": 1, "tac_pham": 1, "mat": 2}


def _cd(ngon_ngu, khoa):
    """Lấy chữ giao diện DNA theo ngôn ngữ (thiếu thì dùng tiếng Việt)."""
    return CHU_DNA.get(ngon_ngu, CHU_DNA["vi"]).get(khoa, CHU_DNA["vi"].get(khoa, khoa))


def fmt_nam(nam, ngon_ngu="vi"):
    """1917 -> '1917' ; -551 -> '551 TCN' (hoặc '551 BC')."""
    if nam is None or nam == "":
        return "?"
    try:
        n = int(nam)
    except (TypeError, ValueError):
        return str(nam)
    return f"{-n} {_cd(ngon_ngu, 'tcn')}" if n < 0 else str(n)


def ten_hien_thi(ten_tg):
    """Tên để hiển thị (VD: 'Confucius' -> 'Khổng Tử (Confucius)')."""
    return DNA_TAC_GIA.get(ten_tg, {}).get("ten_goi") or ten_tg


def chu_viet_tat(ten):
    """'Nam Cao' -> 'NC' — dùng làm ảnh đại diện khi chưa có ảnh chân dung."""
    ten = re.sub(r"\(.*?\)", "", ten or "")
    tu = [t for t in re.split(r"[\s.\-]+", ten) if t]
    if not tu:
        return "?"
    if len(tu) == 1:
        return tu[0][:2].upper()
    return (tu[0][0] + tu[-1][0]).upper()


def sach_cua_tac_gia(ten_tg):
    """Các sách trong thư viện có đúng tên tác giả này."""
    khoa = (ten_tg or "").strip().lower()
    return [s for s in data if (s.get("tac_gia") or "").strip().lower() == khoa]


def _rut_gon(chu, toi_da=170):
    chu = (chu or "").strip()
    return chu if len(chu) <= toi_da else chu[:toi_da].rsplit(" ", 1)[0] + "…"


def lay_moc_tac_gia(ten_tg):
    """Trả về (thông tin tác giả, danh sách mốc đã sắp theo năm)."""
    tt = DNA_TAC_GIA.get(ten_tg)
    if not tt:
        return None, []

    moc = []
    if tt.get("nam_sinh"):
        moc.append({"nam": tt["nam_sinh"], "loai": "doi", "kieu": "sinh", "tieu_de": "Sinh ra",
                    "mo_ta": f"Tại {tt.get('noi_sinh','')}"})

    nam_mat = tt.get("nam_mat")
    for sk in tt.get("su_kien", []):
        moc.append({"nam": sk.get("nam"), "loai": "doi",
                    "kieu": "mat" if nam_mat is not None and sk.get("nam") == nam_mat else "su_kien",
                    "tieu_de": sk.get("tieu_de", ""),
                    "mo_ta": sk.get("mo_ta", ""), "noi": sk.get("noi", "")})

    # ---- tự thêm mốc "Qua đời" nếu phần sự kiện chưa có năm mất ----
    if nam_mat is not None and not any(m.get("kieu") == "mat" for m in moc):
        moc.append({"nam": nam_mat, "loai": "doi", "kieu": "mat", "tu_dong": True,
                    "tieu_de": "Qua đời", "mo_ta": ""})

    # ---- liên kết tác phẩm có sẵn trong data ----
    khoa = ten_tg.strip().lower()
    for s in data:
        if (s.get("tac_gia") or "").strip().lower() == khoa:
            moc.append({"nam": s.get("nam"), "loai": "tac_pham", "kieu": "tac_pham",
                        "tieu_de": s.get("ten_sach", ""),
                        "mo_ta": _rut_gon(s.get("tom_tat")),
                        "chu_de": s.get("chu_de") or [],
                        "so_trang": s.get("so_trang")})

    moc.sort(key=lambda m: (m.get("nam") or 0, THU_TU_MOC.get(m.get("kieu"), 1)))
    return tt, moc


def ban_do_gen(ds_sach, top=6):
    """[(chủ đề, % số sách mang chủ đề đó)] — 'bộ gen' sáng tác của tác giả."""
    if not ds_sach:
        return []
    dem = Counter(t for s in ds_sach for t in (s.get("chu_de") or []))
    return [(t, round(sl * 100 / len(ds_sach))) for t, sl in dem.most_common(top)]


def tac_gia_cung_gen(ten_tg, top=5):
    """Các tác giả (có hồ sơ DNA) trùng nhiều chủ đề nhất với tác giả đang xem."""
    def bo_gen(ten):
        return {t for s in sach_cua_tac_gia(ten) for t in (s.get("chu_de") or [])}

    goc = bo_gen(ten_tg)
    if not goc:
        return []
    nhom = DNA_TAC_GIA.get(ten_tg, {}).get("nhom")
    ket_qua = []
    for ten, tt in DNA_TAC_GIA.items():
        if ten == ten_tg:
            continue
        gen = bo_gen(ten)
        chung = goc & gen
        if not chung:
            continue
        diem = len(chung) / len(goc | gen) + (0.15 if tt.get("nhom") == nhom else 0)
        ket_qua.append((diem, ten, sorted(chung)))
    ket_qua.sort(key=lambda x: -x[0])
    return [(ten, chung) for _, ten, chung in ket_qua[:top]]


def _anh_dai_dien(ten_tg, tt, lop=""):
    """Khung ảnh tròn: chữ viết tắt + ảnh (link 'anh' hoặc JS tự lấy từ Wikipedia qua 'wiki')."""
    if tt.get("anh"):
        img = f"<img class='dna-anh' src='{html.escape(tt['anh'])}' alt=''>"
    elif tt.get("wiki"):
        img = f"<img class='dna-anh' data-wiki='{html.escape(tt['wiki'])}' alt=''>"
    else:
        img = ""
    tat = html.escape(chu_viet_tat(ten_hien_thi(ten_tg)))
    return f"<span class='dna-mat {lop}'><b class='dna-tat'>{tat}</b>{img}</span>"


def render_dna_banner(ngon_ngu="vi"):
    """Băng tiêu đề trang DNA + chuỗi xoắn kép chuyển động."""
    c = lambda k: _cd(ngon_ngu, k)
    xoan = "".join(f"<i style='--i:{i}'></i>" for i in range(22))
    mo_ta = c("banner_mo_ta").format(n=len(DNA_TAC_GIA), s=len(data))
    return f"""
    <div class="dna-banner">
      <div class="dna-banner-chu">
        <span class="dna-kicker">🧬 {c('banner_kicker')}</span>
        <h1>{c('banner_tieu_de')}</h1>
        <p>{html.escape(mo_ta)}</p>
      </div>
      <div class="dna-xoan" aria-hidden="true">{xoan}</div>
    </div>"""


def render_dna(ten_tg, ngon_ngu="vi"):
    ngon_ngu = ngon_ngu if ngon_ngu in CHU_DNA else "vi"
    c = lambda k: _cd(ngon_ngu, k)
    e = html.escape

    tt, moc = lay_moc_tac_gia(ten_tg) if ten_tg else (None, [])
    if not tt:
        return ("<div class='dna-wrap dna-trong'><div class='dna-trong-icon'>🧬</div>"
                f"<p>{c('chua_co')}</p></div>")

    ten = ten_hien_thi(ten_tg)
    ds_sach = sach_cua_tac_gia(ten_tg)
    gen = ban_do_gen(ds_sach)
    ns, nm = tt.get("nam_sinh"), tt.get("nam_mat")
    uoc = "~" if tt.get("uoc_luong") else ""

    # ---------------- 1. PHẦN ĐẦU (HERO) ----------------
    nam_doi = f"{uoc}{fmt_nam(ns, ngon_ngu)} – {uoc + fmt_nam(nm, ngon_ngu) if nm is not None else c('nay')}"
    dong_phu = " · ".join(x for x in [e(tt.get("ten_that", "")), nam_doi,
                                      f"📍 {e(tt['noi_sinh'])}" if tt.get("noi_sinh") else ""] if x)
    pc = "".join(f"<span>{e(p)}</span>" for p in tt.get("phong_cach", []))
    kicker = f"{c('ho_so')} · {e(tt.get('quoc_gia', ''))}".rstrip(" ·")

    tuoi = None
    if ns is not None:
        ket_thuc = nm if nm is not None else NAM_NAY
        tuoi = ket_thuc - ns - (1 if ns < 0 < ket_thuc else 0)      # không có "năm 0"
    o_so = [
        (tuoi, c("tuoi") if nm is not None else c("tuoi_nay"), True),
        (len(ds_sach), c("so_tp"), True),
        (len(moc), c("so_moc"), True),
        (gen[0][0] if gen else "—", c("gen_troi"), False),
    ]
    the_so = ""
    for i, (gia_tri, nhan, dem) in enumerate(o_so):
        gia_tri = "—" if gia_tri is None else gia_tri
        thuoc_tinh = f" data-dem='{gia_tri}'" if dem and isinstance(gia_tri, int) else ""
        the_so += (f"<div class='dna-o' style='--d:{i}'><b{thuoc_tinh}>{e(str(gia_tri))}</b>"
                   f"<span>{e(nhan)}</span></div>")

    # ---- Dải "toàn cảnh cuộc đời": mỗi mốc là 1 chấm trên thanh ----
    cac_nam = [m["nam"] for m in moc if isinstance(m.get("nam"), int)]
    dau = min(cac_nam + ([ns] if ns is not None else [])) if cac_nam or ns is not None else 0
    cuoi = max(cac_nam + [nm if nm is not None else NAM_NAY])
    khoang = max(1, cuoi - dau)
    cham = ""
    for i, m in enumerate(moc):
        n = m.get("nam")
        if not isinstance(n, int):
            continue
        x = (n - dau) * 100 / khoang
        lop = "tp" if m["loai"] == "tac_pham" else m.get("kieu", "su_kien")
        goi_y = e(f"{fmt_nam(n, ngon_ngu)} · {m.get('tieu_de', '')}", quote=True)
        cham += f"<i class='dna-cham {lop}' style='left:{x:.2f}%;--d:{i}' title='{goi_y}'></i>"
    nhan_cuoi = c("nay") if nm is None and cuoi == NAM_NAY else fmt_nam(cuoi, ngon_ngu)

    hero = f"""
    <section class="dna-hero">
      <div class="dna-avatar"><span class="dna-vong"></span>{_anh_dai_dien(ten_tg, tt)}</div>
      <div class="dna-hero-chu">
        <span class="dna-kicker">{kicker}</span>
        <h2 class="dna-ten">{e(ten)}</h2>
        <p class="phu">{dong_phu}</p>
        <p class="gt">{e(tt.get('gioi_thieu', ''))}</p>
        <div class="dna-pc">{pc}</div>
      </div>
      <div class="dna-so">{the_so}</div>
      <div class="dna-doi">
        <div class="dna-doi-nhan">{c('toan_canh')}</div>
        <div class="dna-doi-ray"><i class="dna-doi-fill"></i>{cham}</div>
        <div class="dna-doi-moc"><span>{uoc}{fmt_nam(dau, ngon_ngu)}</span><span>{nhan_cuoi}</span></div>
      </div>
    </section>"""

    # ---------------- 2. BẢN ĐỒ GEN + TÁC GIẢ CÙNG GEN ----------------
    if gen:
        dong_gen = "".join(
            f"<div class='dna-gen-dong' style='--p:{p}%;--mau:{MAU_GEN[i % len(MAU_GEN)]};--d:{i}'>"
            f"<span class='ten'>{e(t)}</span><span class='thanh'><i></i></span><b>{p}%</b></div>"
            for i, (t, p) in enumerate(gen))
    else:
        dong_gen = f"<p class='dna-rong'>{c('chua_co_sach')}</p>"

    lien_quan = tac_gia_cung_gen(ten_tg)
    if lien_quan:
        dong_lq = ""
        for i, (ten2, chung) in enumerate(lien_quan):
            dong_lq += (f"<button class='dna-ban' data-tg='{e(ten2, quote=True)}' style='--d:{i}'>"
                        f"{_anh_dai_dien(ten2, DNA_TAC_GIA[ten2], 'nho')}"
                        f"<span class='dna-ban-chu'><strong>{e(ten_hien_thi(ten2))}</strong>"
                        f"<small>{e(', '.join(chung[:3]))}</small></span>"
                        f"<span class='dna-ban-mui'>→</span></button>")
    else:
        dong_lq = f"<p class='dna-rong'>{c('khong_lq')}</p>"

    luoi = f"""
    <section class="dna-luoi">
      <div class="dna-khoi dna-gen"><h4>🧬 {c('gen')}</h4><p class="phu">{c('gen_phu')}</p>{dong_gen}</div>
      <div class="dna-khoi dna-lienquan"><h4>🔗 {c('cung_gen')}</h4><p class="phu">{c('cung_gen_phu')}</p>{dong_lq}</div>
    </section>"""

    # ---------------- 3. BỘ LỌC + DÒNG THỜI GIAN ----------------
    so_sk = sum(1 for m in moc if m["loai"] != "tac_pham")
    so_tp = len(moc) - so_sk
    loc = f"""
    <div class="dna-loc">
      <span class="dna-loc-tieu">⏳ {c('hanh_trinh')}</span>
      <button class="on" data-loc="all">{c('tat_ca')}<em>{len(moc)}</em></button>
      <button data-loc="sk">✦ {c('loc_sk')}<em>{so_sk}</em></button>
      <button data-loc="tp">📖 {c('loc_tp')}<em>{so_tp}</em></button>
    </div>"""

    hang = ""
    for idx, m in enumerate(moc):
        la_tp = m["loai"] == "tac_pham"
        kieu = m.get("kieu", "su_kien")
        ben  = "phai" if idx % 2 else "trai"
        lop  = "tp" if la_tp else "sk"
        mau  = MAU_MOC.get(kieu, MAU_MOC["su_kien"])

        tieu_de, mo_ta = m.get("tieu_de", ""), m.get("mo_ta", "")
        if kieu == "sinh":
            icon, nhan = "🌱", c("sinh_nhan")
            tieu_de = c("sinh")
            mo_ta = f"{c('tai')} {tt.get('noi_sinh', '')}" if tt.get("noi_sinh") else ""
        elif kieu == "mat":
            icon, nhan = "🕯️", c("mat_nhan")
            if m.get("tu_dong"):
                tieu_de = c("mat")
        elif la_tp:
            icon, nhan = "📖", c("tac_pham")
        else:
            icon, nhan = "✦", c("su_kien")

        phu = ""
        if la_tp and m.get("chu_de"):
            chips = "".join(f"<span class='dna-chip'>{e(str(t))}</span>" for t in m["chu_de"])
            phu = f"<div class='dna-tags'>{chips}</div>"
        elif m.get("noi"):
            phu = f"<span class='dna-noi'>📍 {e(m['noi'])}</span>"

        chan = ""
        if la_tp:
            bits = [f"📄 {m['so_trang']} {c('trang')}"] if m.get("so_trang") else []
            bits.append(f"✓ {c('trong_tv')}")
            chan = f"<div class='dna-chan'>{e(' · '.join(bits))}</div>"

        n = m.get("nam")
        if isinstance(n, int) and n < 0:
            nut = f"<b>{-n}</b><small>{c('tcn')}</small>"
        else:
            nut = f"<b>{n if n is not None else '?'}</b>"

        hang += f"""
        <div class="dna-moc {ben} {lop} {kieu}" style="--mau:{mau}">
          <i class="dna-ket"></i>
          <div class="dna-nut {lop}">{nut}</div>
          <div class="dna-the">
            <div class="dna-the-dau"><span class="nhan">{icon} {e(nhan)}</span><span class="dna-nam">{uoc}{e(fmt_nam(n, ngon_ngu))}</span></div>
            <h3>{e(tieu_de)}</h3>
            {f"<p>{e(mo_ta)}</p>" if mo_ta else ""}
            {phu}{chan}
          </div>
        </div>"""

    dong_thoi_gian = f"<div class='dna-line'><div class='dna-ray'><i class='dna-sao'></i>{hang}</div></div>"
    nguon = f"<p class='dna-nguon'>{c('nguon')}</p>"
    return f"<div class='dna-wrap' data-loc='all'>{hero}{luoi}{loc}{dong_thoi_gian}{nguon}</div>"


# ---------------- ĐIỀU KHIỂN TRANG DNA (lọc nhóm, trước/sau, ngẫu nhiên) ----------------
def lua_chon_dna(nhom="tat_ca", ngon_ngu="vi"):
    """Danh sách (nhãn, giá trị) cho Dropdown chọn tác giả, lọc theo nhóm."""
    ds = [t for t, tt in DNA_TAC_GIA.items() if nhom in (None, "", "tat_ca") or tt.get("nhom") == nhom]
    ds.sort(key=lambda t: bo_dau(ten_hien_thi(t)))
    ket_qua = []
    for t in ds:
        tt = DNA_TAC_GIA[t]
        nam = fmt_nam(tt.get("nam_sinh"), ngon_ngu)
        if tt.get("nam_mat") is not None:
            nam += "–" + fmt_nam(tt["nam_mat"], ngon_ngu)
        ket_qua.append((f"{ten_hien_thi(t)} · {nam}", t))
    return ket_qua


def lua_chon_nhom_dna(ngon_ngu="vi"):
    """Nhãn các nhóm kèm số tác giả, VD: 'Việt Nam · 18'."""
    cot = 1 if ngon_ngu == "en" else 0
    ket_qua = []
    for khoa, nhan in NHOM_DNA.items():
        dem = len(DNA_TAC_GIA) if khoa == "tat_ca" else sum(1 for tt in DNA_TAC_GIA.values() if tt.get("nhom") == khoa)
        if dem:
            ket_qua.append((f"{nhan[cot]} · {dem}", khoa))
    return ket_qua


def doi_nhom_dna(nhom, hien_tai, ngon_ngu="vi"):
    lua_chon = lua_chon_dna(nhom, ngon_ngu)
    gia_tri = [v for _, v in lua_chon]
    chon = hien_tai if hien_tai in gia_tri else (gia_tri[0] if gia_tri else None)
    return gr.update(choices=lua_chon, value=chon)


def dna_ke_ben(hien_tai, nhom, buoc):
    ds = [v for _, v in lua_chon_dna(nhom)]
    if not ds:
        return gr.update()
    i = ds.index(hien_tai) if hien_tai in ds else -1
    return gr.update(value=ds[(i + buoc) % len(ds)])


def dna_ngau_nhien(hien_tai, nhom):
    ds = [v for _, v in lua_chon_dna(nhom) if v != hien_tai]
    return gr.update(value=random.choice(ds)) if ds else gr.update()


def dna_tu_cau_noi(gia_tri, ngon_ngu="vi"):
    """Bấm 1 tác giả 'cùng gen' trong HTML -> JS ghi 'Tên|thời gian' vào ô ẩn -> mở hồ sơ đó."""
    ten = (gia_tri or "").split("|")[0]
    if ten not in DNA_TAC_GIA:
        return gr.update(), gr.update()
    return gr.update(value="tat_ca"), gr.update(choices=lua_chon_dna("tat_ca", ngon_ngu), value=ten)
```

```python
################################################################################

# MODEL AI

################################################################################

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

with open("books_300.json", "r", encoding="utf-8") as f:
    sach = json.load(f)
print(f"Đã tải {len(sach)} cuốn sách.")

model = SentenceTransformer("AITeamVN/Vietnamese_Embedding")

van_ban = [
    (
        f"Tên sách: {s['ten_sach']}. "
        f"Tác giả: {s['tac_gia']}. "
        f"Chủ đề: {', '.join(s['chu_de'])}. "
        f"Tóm tắt: {s['tom_tat']}"
    )
    for s in sach
]
vector_sach = model.encode(van_ban, convert_to_tensor=True)

tat_ca_chu_de = set()
for s in sach:
    tat_ca_chu_de.update(s["chu_de"])

def phan_tich_cau_hoi(cau_hoi):
    prompt = f"""
Bạn là hệ thống phân tích câu hỏi cho chatbot gợi ý sách.

Nhiệm vụ:
Đọc câu hỏi của người dùng và trích xuất thông tin dưới dạng JSON.

Chỉ trả về JSON hợp lệ.

Schema:

{{
  "intent": "",
  "author": "",
  "book_title": "",
  "topics": [],
  "keywords": [],
  "filters": {{
      "year": null,
      "max_pages": null
  }}
}}

Giải thích các trường:

- intent:
    "recommend" nếu người dùng muốn được gợi ý sách
    "search_author" nếu muốn tìm theo tác giả
    "search_book" nếu muốn tìm theo tên sách

- author:
    tên tác giả nếu có

- book_title:
    tên sách nếu có

- topics:
    danh sách chủ đề chính

- keywords:
    các từ khóa quan trọng phục vụ tìm kiếm

- emotion:
    cảm xúc của người dùng nếu có
    Ví dụ:
    buồn
    cô đơn
    chữa lành
    truyền cảm hứng
    vui vẻ

Nếu không có thì để chuỗi rỗng hoặc mảng rỗng.

Câu hỏi:

{cau_hoi}
"""


    try:
        response = client.models.generate_content(
            model="gemini-flash-latest",
            contents=prompt,
            config=types.GenerateContentConfig(
                temperature=0,
                response_mime_type="application/json",
            ),
        )

        print("===== RESPONSE ===telek")
        print(response.text)
        print("====================")

        return json.loads(response.text)

    except Exception as e:
        traceback.print_exc()

        if "429" in str(e) or "RESOURCE_EXHAUSTED" in str(e):
            return "QUOTA_HET"

        return None
def loc_theo_tac_gia(ds_ket_qua, tac_gia):

    if not tac_gia:
        return ds_ket_qua

    ket_qua = []

    tac_gia = tac_gia.lower().strip()

    for book, score in ds_ket_qua:

        ten_db = book["tac_gia"].lower()

        if tac_gia in ten_db or ten_db in tac_gia:
            ket_qua.append((book, score))

    return ket_qua

# def phat_hien_tac_gia(cau_hoi):
#     cau_hoi_lower = cau_hoi.lower()
#     da_them = set()
#     tim_thay = []  # list các tuple (vị trí trong câu hỏi, tên gốc)
#     for s in sach:
#         ten = s["tac_gia"].lower()
#         if ten in cau_hoi_lower and ten not in da_them:
#             vi_tri = cau_hoi_lower.index(ten)
#             tim_thay.append((vi_tri, s["tac_gia"]))
#             da_them.add(ten)
#     tim_thay.sort(key=lambda x: x[0])  # sắp theo thứ tự xuất hiện trong câu hỏi
#     return [ten for _, ten in tim_thay]

def semantic_search(query, top_k=10):

    vector_query = model.encode(
        query,
        convert_to_tensor=True
    )

    scores = util.cos_sim(
        vector_query,
        vector_sach
    )[0]

    top = scores.topk(
        k=min(top_k, len(sach))
    )

    ket_qua = []

    for score, idx in zip(top.values, top.indices):

        ket_qua.append(
            (
                sach[int(idx)],
                float(score)
            )
        )

    return ket_qua

def book_to_context(s):
    return f"""
Tên sách: {s['ten_sach']}
Tác giả: {s['tac_gia']}
Năm xuất bản: {s['nam']}
Chủ đề: {', '.join(s['chu_de'])}
Tóm tắt: {s['tom_tat']}
""".strip()
def generate_answer(question, books, ngon_ngu="vi"):      # ngon_ngu: theo phần Cài đặt

    if not books:
        return _chu("ai_khong_thay", ngon_ngu)   # "Xin lỗi, mình không tìm thấy cuốn sách phù hợp trong cơ sở dữ liệu."

    context = "\n\n".join(
        book_to_context(book)
        for book in books
    )

    prompt = f"""
Bạn là một chatbot tư vấn sách.

Nhiệm vụ của bạn là trả lời người dùng dựa DUY NHẤT trên danh sách sách được cung cấp.

=====================
CÂU HỎI NGƯỜI DÙNG
=====================

{question}

=====================
DANH SÁCH SÁCH
=====================

{context}

=====================
QUY TẮC
=====================

1. Chỉ sử dụng thông tin trong DANH SÁCH SÁCH.

2. Không được tự ý thêm bất kỳ cuốn sách nào ngoài danh sách.

3. Nếu không có cuốn nào thật sự phù hợp, hãy nói rõ rằng cơ sở dữ liệu hiện tại chưa có.

4. Nếu có nhiều sách phù hợp, hãy ưu tiên giới thiệu 2-3 cuốn nổi bật và giải thích ngắn gọn lý do.

5. {_chu("ai_quy_tac_5", ngon_ngu)}

"""

    try:

        response = client.models.generate_content(
            model="gemini-flash-latest",
            contents=prompt,
            config=types.GenerateContentConfig(
                temperature=0.3
            )
        )

        return response.text

    except Exception as e:

        print(e)

        return _chu("ai_loi", ngon_ngu)          # "Đã xảy ra lỗi khi tạo câu trả lời."
def tra_loi(cau_hoi, ngon_ngu="vi"):
    # Bước 1: Phân tích câu hỏi bằng Gemini
    parsed = phan_tich_cau_hoi(cau_hoi)

    if parsed == "QUOTA_HET":
        return _chu("ai_het_quota", ngon_ngu)    # "Hệ thống đang tạm hết lượt sử dụng miễn phí trong hôm nay..."

    if parsed is None:
        return _chu("ai_khong_hieu", ngon_ngu)   # "Xin lỗi, mình không hiểu câu hỏi của bạn."

    # Bước 2: Tạo câu truy vấn cho Semantic Search
    query = " ".join(parsed.get("keywords", []))

    # Bước 3: Tìm kiếm bằng Embedding
    ket_qua = semantic_search(query)

    # Bước 4: Lọc theo tác giả (nếu có)
    ket_qua = loc_theo_tac_gia(
    ket_qua,
    parsed.get("author", "")
    )
    if not ket_qua:
        return _chu("ai_khong_sach", ngon_ngu)   # "Không tìm thấy sách phù hợp."

    # Bước 5: Chỉ lấy danh sách sách (bỏ điểm cosine)
    books = [book for book, score in ket_qua]

    # Bước 6: Sinh câu trả lời bằng Gemini
    return generate_answer(
        cau_hoi,
        books,
        ngon_ngu
    )
```

```python
# HÀM XỬ LÝ LƯU TRỮ SÁCH (JSON)


DUONG_DAN_JSON = "du_lieu_sach.json"


def doc_du_lieu_sach():
    """Đọc toàn bộ danh sách sách từ file JSON, trả về list[dict]."""
    if not os.path.exists(DUONG_DAN_JSON):
        return []
    try:
        with open(DUONG_DAN_JSON, "r", encoding="utf-8") as f:
            du_lieu = json.load(f)
        return du_lieu.get("sach", [])
    except (json.JSONDecodeError, FileNotFoundError):
        return []


def ghi_du_lieu_sach(danh_sach):
    """Ghi list[dict] sách vào file JSON."""
    with open(DUONG_DAN_JSON, "w", encoding="utf-8") as f:
        json.dump({"sach": danh_sach}, f, ensure_ascii=False, indent=2)


def _sang_dataframe(danh_sach, ngon_ngu="vi"):
    """Chuyển list[dict] -> dữ liệu bảng gr.Dataframe (kèm tiêu đề cột theo ngôn ngữ).
    Gradio 6 không đổi tiêu đề qua gr.update(headers=...) nên gửi kèm "headers" trong value."""
    nhan_tt = dict(zip(TU_DIEN["lt_ds_trang_thai"]["gia_tri"], _chu("lt_ds_trang_thai", ngon_ngu)))
    return {
        "headers": _chu("lt_cot", ngon_ngu),
        "data": [
            [s["ten_sach"], s["trang_hien_tai"], nhan_tt.get(s["trang_thai"], s["trang_thai"]), s["yeu_thich"]]
            for s in danh_sach
        ],
    }


def _sang_dropdown_choices(danh_sach):
    """Lấy danh sách tên sách để hiển thị trong Dropdown cập nhật."""
    return [s["ten_sach"] for s in danh_sach]


def hien_thi_tat_ca_sach(ngon_ngu="vi"):
    """Trả về toàn bộ danh sách sách (dùng khi mở trang Lưu trữ)."""
    danh_sach = doc_du_lieu_sach()
    return (
        _sang_dataframe(danh_sach, ngon_ngu),
        gr.update(choices=_sang_dropdown_choices(danh_sach), value=None)
    )


def luu_sach_moi(ten, trang, trang_thai, yeu_thich, ngon_ngu="vi"):
    """Thêm 1 cuốn sách mới vào kho lưu trữ."""
    ten = (ten or "").strip()
    danh_sach = doc_du_lieu_sach()

    if ten == "":
        gr.Warning("Vui lòng nhập tên sách trước khi lưu!")
        return (
            _sang_dataframe(danh_sach, ngon_ngu),
            gr.update(choices=_sang_dropdown_choices(danh_sach)),
            gr.update(), gr.update(), gr.update(), gr.update()
        )

    if any(s["ten_sach"].lower() == ten.lower() for s in danh_sach):
        gr.Warning(f"Sách '{ten}' đã tồn tại trong kho lưu trữ!")
        return (
            _sang_dataframe(danh_sach, ngon_ngu),
            gr.update(choices=_sang_dropdown_choices(danh_sach)),
            gr.update(), gr.update(), gr.update(), gr.update()
        )

    danh_sach.append({
        "ten_sach": ten,
        "trang_hien_tai": int(trang) if trang is not None else 0,
        "trang_thai": trang_thai,
        "yeu_thich": bool(yeu_thich),
    })
    ghi_du_lieu_sach(danh_sach)

    # Xoá trắng các ô nhập sau khi lưu thành công
    return (
        _sang_dataframe(danh_sach, ngon_ngu),
        gr.update(choices=_sang_dropdown_choices(danh_sach)),
        "", 0, "Chưa đọc", False
    )


def loc_sach(trang_thai_muon_loc, chi_yeu_thich, ngon_ngu="vi"):
    """Lọc sách theo trạng thái (hoặc tất cả) và tùy chọn chỉ hiện yêu thích."""
    danh_sach = doc_du_lieu_sach()

    if trang_thai_muon_loc is not None:
        danh_sach = [s for s in danh_sach if s["trang_thai"] == trang_thai_muon_loc]

    if chi_yeu_thich:
        danh_sach = [s for s in danh_sach if s.get("yeu_thich")]

    return _sang_dataframe(danh_sach, ngon_ngu)


def tai_thong_tin_sach(ten_sach_chon):
    """Khi chọn 1 sách trong Dropdown cập nhật, tự động điền thông tin hiện có."""
    if not ten_sach_chon:
        return 0, "Chưa đọc", False

    danh_sach = doc_du_lieu_sach()
    for s in danh_sach:
        if s["ten_sach"] == ten_sach_chon:
            return s["trang_hien_tai"], s["trang_thai"], s["yeu_thich"]

    return 0, "Chưa đọc", False


def cap_nhat_tien_do(ten_sach_chon, trang_moi, trang_thai_moi, yeu_thich_moi, ngon_ngu="vi"):
    """Cập nhật trang hiện tại / trạng thái / yêu thích cho 1 sách đã chọn."""
    danh_sach = doc_du_lieu_sach()

    if not ten_sach_chon:
        gr.Warning("Vui lòng chọn sách cần cập nhật!")
        return _sang_dataframe(danh_sach, ngon_ngu), gr.update(choices=_sang_dropdown_choices(danh_sach))

    da_tim_thay = False
    for s in danh_sach:
        if s["ten_sach"] == ten_sach_chon:
            s["trang_hien_tai"] = int(trang_moi) if trang_moi is not None else s["trang_hien_tai"]
            s["trang_thai"] = trang_thai_moi
            s["yeu_thich"] = bool(yeu_thich_moi)
            da_tim_thay = True
            break

    if da_tim_thay:
        ghi_du_lieu_sach(danh_sach)
        gr.Info(f"Đã cập nhật '{ten_sach_chon}'!")
    else:
        gr.Warning("Không tìm thấy sách trong kho lưu trữ!")

    return _sang_dataframe(danh_sach, ngon_ngu), gr.update(choices=_sang_dropdown_choices(danh_sach))


def xoa_sach_khoi_kho(ten_sach_chon, ngon_ngu="vi"):
    """Xoá 1 sách khỏi kho lưu trữ."""
    danh_sach = doc_du_lieu_sach()

    if not ten_sach_chon:
        gr.Warning("Vui lòng chọn sách cần xoá!")
        return (
            _sang_dataframe(danh_sach, ngon_ngu),
            gr.update(choices=_sang_dropdown_choices(danh_sach), value=None),
            0, "Chưa đọc", False
        )

    danh_sach_moi = [s for s in danh_sach if s["ten_sach"] != ten_sach_chon]

    if len(danh_sach_moi) == len(danh_sach):
        gr.Warning("Không tìm thấy sách để xoá!")
    else:
        ghi_du_lieu_sach(danh_sach_moi)
        gr.Info(f"Đã xoá '{ten_sach_chon}' khỏi kho lưu trữ!")

    return (
        _sang_dataframe(danh_sach_moi, ngon_ngu),
        gr.update(choices=_sang_dropdown_choices(danh_sach_moi), value=None),
        0, "Chưa đọc", False
    )

# Dựng html
def render(ket_qua, q, chon, ngon_ngu="vi"):
    dang_loc = ""
    if chon:
        chips = " ".join(f"<span style='background:#0d9488;color:#fff;font:11px monospace;"
                         f"padding:2px 8px;border-radius:100px'>{html.escape(str(t))}</span>"
                         for t in sorted(map(str, chon)))
        dang_loc = (f"<div style='margin:0 0 8px;color:var(--the-sach-chu,#334155)'>{_chu('tk_dang_loc', ngon_ngu)} {chips}</div>")

    header = (dang_loc +
              f"<p style='font-family:monospace;color:#0d9488;margin:2px 0 14px'>"
              + _chu("tk_tim_thay", ngon_ngu).format(a=len(ket_qua), b=len(data)) + "</p>")   # "Tìm thấy <b>x</b> / y tác phẩm"

    if not ket_qua:
        return (header +
                "<div style='padding:40px;text-align:center;color:var(--the-sach-phu,#64748b)'>"
                + _chu("tk_khong_co", ngon_ngu) + "</div>")   # "Không có tác phẩm nào khớp..."

    cards = ""
    for i in ket_qua:
        tags = ""
        for t in lay_chu_de(i):
            nb = t in chon
            bg = "#0d9488" if nb else "rgba(13,148,136,.12)"
            fg = "#fff" if nb else "#0d9488"
            tags += (f"<span style='font:11px monospace;background:{bg};color:{fg};"
                     f"padding:3px 9px;border-radius:100px;margin:0 5px 5px 0;"
                     f"display:inline-block'>{html.escape(str(t))}</span>")
        cards += f"""
        <div style="border:1px solid var(--the-sach-vien,#e2e8f0);border-radius:12px;
                    padding:14px 18px;margin-bottom:12px;background:var(--the-sach-nen,#fff)">
          <div style="display:flex;justify-content:space-between;gap:12px">
            <h3 style="margin:0 0 2px;font-size:17px;font-weight:600;color:var(--the-sach-tieu-de,#0f172a)">
              {highlight(i['ten_sach'], q)}</h3>
            <span style="color:var(--the-sach-phu,#64748b);font-size:13px;white-space:nowrap">{i.get('nam','')}</span>
          </div>
          <p style="margin:0 0 8px;color:#0d9488;font-size:14px;font-weight:600">
             ✍ {highlight(i['tac_gia'], q)}</p>
          <p style="margin:0 0 10px;font-size:13.5px;color:var(--the-sach-chu,#475569)">{html.escape(i.get('tom_tat',''))}</p>
          <div>{tags}</div>
        </div>"""
    return header + f"<div style='max-height:460px;overflow-y:auto;padding-right:8px'>{cards}</div>"

def goi_y_ngau_nhien(n=10, ngon_ngu="vi"):
    chon = random.sample(data, min(n, len(data)))

    header = ("<p style='font-family:monospace;color:#0d9488;margin:2px 0 14px'>"
              + _chu("gy_tieu_de", ngon_ngu) + "</p>")   # "📚 Gợi ý cho bạn · một vài tác phẩm ngẫu nhiên từ thư viện"

    cards = ""
    for i in chon:
        tags = ""
        for t in i.get("chu_de", []):
            tags += ("<span style='font:11px monospace;background:rgba(13,148,136,.12);"
                     "color:#0d9488;padding:3px 9px;border-radius:100px;margin:0 5px 5px 0;"
                     "display:inline-block'>" + html.escape(t) + "</span>")
        cards += f"""
        <div style="border:1px solid var(--the-sach-vien,#e2e8f0);border-radius:12px;
                    padding:14px 18px;margin-bottom:12px;background:var(--the-sach-nen,#fff)">
          <div style="display:flex;justify-content:space-between;gap:12px">
            <h3 style="margin:0 0 2px;font-size:17px;font-weight:600;color:var(--the-sach-tieu-de,#0f172a)">
              {html.escape(i['ten_sach'])}</h3>
            <span style="color:var(--the-sach-phu,#64748b);font-size:13px;white-space:nowrap">{i.get('nam','')}</span>
          </div>
          <p style="margin:0 0 8px;color:#0d9488;font-size:14px;font-weight:600">
             ✍ {html.escape(i['tac_gia'])}</p>
          <p style="margin:0 0 10px;font-size:13.5px;color:var(--the-sach-chu,#475569)">{html.escape(i.get('tom_tat',''))}</p>
          <div>{tags}</div>
        </div>"""
    return header + f"<div style='max-height:460px;overflow-y:auto;padding-right:8px'>{cards}</div>"
```

```python
#Frontend

#Theme này chỉ giúp chỉnh màu nút
THEME = gr.themes.Soft(primary_hue="teal", neutral_hue="slate")       #Nút variant primary và neutral


with gr.Blocks(fill_height=True, elem_id="app",css=CSS,theme=THEME, js=JS_DNA) as Library:

    # -------------- KHAI BÁO CÁC BIẾN TẠM THỜI --------------

    #Biến sd trong crop ảnh
    anh_goc = gr.State(None)
    diem_chon = gr.State(None)

    #Biến sd cho lịch sử chat
    luu_tru_chat = gr.State({})       # Dict chứa toàn bộ lịch sử: {"Tên Chat": [nội dung]}
    chat_id_hien_tai = gr.State(None) # Đoạn chat đang mở

    #Biến lưu cài đặt (ngôn ngữ, sáng/tối...) ngay trên trình duyệt -> mở lại app vẫn giữ nguyên
    cai_dat_luu = gr.BrowserState(dict(MAC_DINH_CAI_DAT), storage_key="thu_vien_so_cai_dat",
                                  secret="thu-vien-so-nhom-7")

    with gr.Sidebar(open=True,width=200):
        dich(gr.Markdown("### Tương tác"), value="sb_tuong_tac")
        trang_chu = dich(gr.Button("🏠 Trang chủ"), value="sb_trang_chu")            # 4 nhánh chức năng chính, đặt trang chủ UI mở web
        btn_AI = dich(gr.Button("🤖 Chat AI"), value="sb_chat")
        luu_tru = dich(gr.Button("📚 Lưu trữ"), value="sb_luu_tru")
        dna_btn = dich(gr.Button("🧬 DNA tác giả (Beta)", variant="secondary"), value="sb_dna")
        cai_dat = dich(gr.Button("⚙️ Cài đặt"), value="sb_cai_dat")

# DNA TÁC GIẢ
    with gr.Group(visible=False, elem_id="trang_dna") as Trang_DNA:
        with gr.Column(scale=1):
            # gr.Markdown("# 🧬 DNA TÁC GIẢ")    # (bản cũ) -> thay bằng băng tiêu đề có chuỗi xoắn DNA chuyển động
            dich(gr.HTML(render_dna_banner()), value=render_dna_banner)

            # Lọc nhóm tác giả: Tất cả / Việt Nam / Thế giới / Truyện tranh / Triết học & kỹ năng
            dna_nhom = dich(gr.Radio(choices=lua_chon_nhom_dna(), value="tat_ca", show_label=False,
                                     container=False, elem_classes="cd-chon"), choices=lua_chon_nhom_dna)
            with gr.Row(elem_id="dna_dieu_khien"):
                dna_chon = dich(gr.Dropdown(choices=lua_chon_dna(),
                                       value=next(iter(DNA_TAC_GIA), None),
                                       label="Chọn tác giả", filterable=True, scale=8), label="dna_chon")
                dna_truoc = gr.Button("◀", scale=0, min_width=46)      # tác giả trước
                dna_ngau = gr.Button("🎲", scale=0, min_width=46)      # tác giả ngẫu nhiên
                dna_sau = gr.Button("▶", scale=0, min_width=46)        # tác giả sau
            dna_cau = gr.Textbox(elem_id="dna_cau", show_label=False, container=False)   # ô ẩn: JS gửi tên tác giả "cùng gen"
            dna_html = gr.HTML()

#########################  TRANG CHỦ   #########################

    with gr.Group(visible=True) as Trang_Chu:                 #Vì Group mới add ở đây mà chưa css nên trông trang chủ xem hơi kỳ

      # -------------- MAIN CONTENT --------------                                    #Tùy chỉnh thông tin
      with gr.Column(scale=1):
          # Header
          with gr.Row(elem_classes="header", equal_height=True):
              with gr.Column(scale=1, min_width=120):
                    gr.HTML(
                        "<img src='https://i.ibb.co/bjFJNH6V/Screenshot-2026-07-26-234722.png' "
                        "style='height:56px;width:auto;object-fit:contain;display:block'>"
                    )

              with gr.Column(scale=5):
                  dich(gr.Markdown("""
  #  THƯ VIỆN SỐ - NHÓM 7

  Khám phá • Tra cứu • Theo dõi tiến độ đọc
  """), value="tc_tieu_de")
              with gr.Column(scale=2, min_width=220):
                  dich(gr.Markdown("### *Banner quảng cáo*"), value="tc_banner")


          # -------------- SEARCH SECTION --------------
          with dich(gr.Accordion("🔍 Tìm kiếm", open=False), label="tc_tim_kiem"):
              with gr.Row():
                  with gr.Column(scale=1, min_width=180):

                    #Sắp xếp
                      sap_xep = gr.Dropdown(
                          choices=["Mới nhất", "Cũ nhất", "Tên A–Z"],
                          value="Mới nhất",
                          show_label=False,
                          interactive=True,
                          filterable=False
                      )
                  #Thanh tìm kiếm

                  with gr.Column(scale=8, min_width=200):
                      search = gr.Textbox(
                          placeholder="Nhập Tìm theo tên tác phẩm hoặc tác giả...",
                          show_label=False,
                          interactive=True,
                      )

                      #Nút HÌNH ẢNH
                  anh_btn = gr.Button("📷", elem_id="anh_btn",
                                    scale=1, min_width=56, variant="secondary"
                                    )

              loc_btn = gr.Button("⚙  Bộ lọc thể loại chi tiết", variant="secondary")

              # Đăng ký đổi ngôn ngữ cho khu tìm kiếm
              dich(sap_xep, choices="tc_sap_xep")
              dich(search, placeholder="tc_o_tim")
              dich(loc_btn, value="tc_loc_btn")

          # -------------- MAIN PANEL --------------
          with gr.Group() as content:
            with gr.Column(elem_classes="main-panel"):
                dich(gr.Markdown("## Trang chủ"), value="tc_trang_chu")

                # Hàng thẻ lựa chọn
                with gr.Row(equal_height=True, elem_classes="main-row"):
                    for i in range(1, 5):
                        with gr.Column(scale=1, min_width=0):
                            gr.Image(
                                value=None,
                                height=180,
                                show_label=False,
                                container=False,
                                interactive=False,
                                elem_classes="card-image"
                            )

                            dich(gr.Markdown(
                                f"**Lựa chọn {i}**",
                                elem_classes="card-label"
                            ), value=lambda l, i=i: f"**{_chu('tc_lua_chon', l)} {i}**")

                    gr.Markdown("---")

                with gr.Column(visible=True, elem_id="goi_y_box") as goi_y_box:
                    goi_y_html = dich(gr.HTML(goi_y_ngau_nhien()), value=lambda l: goi_y_ngau_nhien(ngon_ngu=l))

                output_trang_chu = gr.HTML(visible=False)
    # --------------  Các Pop-up --------------

    # Pop up bộ lọc
    with gr.Column(visible=False, elem_id="modal_box") as popup_loc:
        with gr.Row(equal_height=True):
            dich(gr.Markdown("## 🏷️ Thể loại"), value="pl_the_loai")
            dong_loc_btn = gr.Button("✕", scale=0, min_width=44)

        dich(gr.Markdown(f"*Chọn một hoặc nhiều chủ đề ({len(CHU_DE)} chủ đề):*"),
             value=lambda l: _chu("pl_huong_dan", l).format(n=len(CHU_DE)))

        chu_de_box = gr.CheckboxGroup(
            choices=CHU_DE,
            value=[],
            show_label=False,
            elem_id="chu_de_box",
            interactive=True,
            elem_classes="pill-tags"
        )

        with gr.Row():
            xoa_btn = gr.Button("Bỏ chọn tất cả", scale=0, min_width=150)
            gr.HTML("")  # spacer
            tim_btn = gr.Button("Tìm kiếm", variant="primary", scale=0, min_width=150)
            dich(xoa_btn, value="pl_bo_chon")
            dich(tim_btn, value="pl_tim")


    # Pop up hình ảnh
    with gr.Column(visible=False, elem_id="img_modal_box") as popup_anh:
      with gr.Row():
          dich(gr.Markdown("## Tìm bằng ảnh"), value="pa_tieu_de")
          dong_anh_btn = gr.Button("✕", scale=0, min_width=44) # NOTE: Nút đóng cửa sổ pop-up hình ảnh

      dich(gr.Markdown("Tải ảnh lên → dùng công cụ **cắt (crop)** để khoanh vùng chữ cần đọc → bấm **Đọc chữ**."), value="pa_huong_dan")

            # Khung tải ảnh lên hoặc dán ảnh từ clipboard
      img_input = gr.Image(
            type="pil",
            show_label=False,
            elem_id="img_editor",
            sources=["upload", "clipboard"],
            height=340
      )

      # Hàng chứa nút xoá vùng chọn và thông báo trạng thái thao tác ảnh
      with gr.Row():
          xoa_vung_btn = gr.Button("Xoá vùng chọn", variant="secondary") # NOTE: Nút để reset lại vùng chọn trên ảnh
          vung_status = gr.Markdown("*Chưa chọn vùng — sẽ đọc toàn bộ ảnh.*") # NOTE: Nhãn hiển thị trạng thái chọn vùng hiện tại

      with gr.Row():
          gr.HTML("")  # spacer (khoảng trống canh lề)
          ocr_btn = gr.Button("Đọc chữ → thanh tìm kiếm", variant="primary", scale=0, min_width=240) # NOTE: Nút kích hoạt tiến trình trích xuất văn bản OCR
          dich(xoa_vung_btn, value="pa_xoa_vung")
          dich(ocr_btn, value="pa_ocr")

######################### CHATBOT #########################

    with gr.Group(visible=False, elem_id="chatbox-container") as Chatbox:
        chat = gr.Chatbot(elem_id="chatbot",                               #editable="user", buttons=[] ẩn nút share và copy
                          show_label=False, buttons=[] ,
                          placeholder="## Bạn cứ hỏi đi! 👋",
                          )
        # -------------- Textbox  --------------                                           #Ko tới thì chỉnh lại đc
        with gr.Row():
            o_nhap = gr.Textbox(
                placeholder="Ví dụ: Sách của Nguyễn Nhật Ánh hoặc sách về kinh doanh...",
                show_label=False,
                scale=6
            )
            nut_gui = gr.Button("Gửi", scale=1)
            dich(chat, placeholder="ch_placeholder")
            dich(o_nhap, placeholder="ch_o_nhap")
            dich(nut_gui, value="ch_gui")


    # --------------  Sidebar của Chatbot --------------
        with gr.Sidebar(position="right", open=False,width=220):                #Ko chỉnh đc độ lớn, phải css
            gr.Markdown("### AI ASSISTANT")
            btn_them_moi=dich(gr.Button("+ Thêm mới"), value="ch_them_moi")
            dich(gr.Markdown("Gần đây"), value="ch_gan_day")

            # Danh sách lịch sử
            danh_sach_lich_su = gr.Radio(
                choices=[],
                value=None,
                show_label=False,
                elem_classes="lich-su-radio",
                interactive=True
            )

######################### LƯU TRỮ #########################
    with gr.Group(visible=False) as Luu_Tru:
      dich(gr.Markdown("## 📚 Lưu trữ sách của bạn"), value="lt_tieu_de")

      # -------------- THÊM SÁCH MỚI --------------
      with dich(gr.Accordion("➕ Thêm sách mới", open=False), label="lt_them"):
          with gr.Row():
              ten_sach_moi = gr.Textbox(
                  label="Tên sách",
                  placeholder="Nhập tên sách...",
                  scale=3
              )
              trang_hien_tai_moi = gr.Number(
                  label="Trang hiện tại",
                  value=0,
                  minimum=0,
                  scale=1
              )
              trang_thai_moi = gr.Dropdown(
                  label="Trạng thái",
                  choices=["Chưa đọc", "Đang đọc", "Đã đọc xong"],
                  value="Chưa đọc",
                  scale=1
              )

          with gr.Row():
              yeu_thich_moi = gr.Checkbox(
                  label="⭐ Đánh dấu yêu thích",
                  value=False
              )
              luu_sach_btn = gr.Button("💾 Lưu sách", variant="primary", scale=0, min_width=150)

          # Đăng ký đổi ngôn ngữ (giá trị trạng thái vẫn lưu tiếng Việt, chỉ đổi nhãn hiển thị)
          dich(ten_sach_moi, label="lt_ten_sach", placeholder="lt_nhap_ten")
          dich(trang_hien_tai_moi, label="lt_trang")
          dich(trang_thai_moi, label="lt_trang_thai", choices="lt_ds_trang_thai")
          dich(yeu_thich_moi, label="lt_danh_dau")
          dich(luu_sach_btn, value="lt_luu")

      # -------------- BỘ LỌC NHANH --------------
      with gr.Row():
          tat_ca_btn = gr.Button("📋 Tất cả")
          chua_doc_btn = gr.Button("📕 Chưa đọc")
          dang_doc_btn = gr.Button("📖 Đang đọc")
          da_xong_btn = gr.Button("✅ Đã đọc xong")
          yeu_thich_checkbox = gr.Checkbox(label="⭐ Chỉ hiện yêu thích", value=False)
          dich(tat_ca_btn, value="lt_tat_ca"); dich(chua_doc_btn, value="lt_chua_doc")
          dich(dang_doc_btn, value="lt_dang_doc"); dich(da_xong_btn, value="lt_da_xong")
          dich(yeu_thich_checkbox, label="lt_chi_yt")

      gr.Markdown("---")

      # -------------- DANH SÁCH SÁCH --------------
      with gr.Group():
          danh_sach_sach = gr.Dataframe(
              headers=["Tên sách", "Trang hiện tại", "Trạng thái", "Yêu thích"],
              datatype=["str", "number", "str", "bool"],
              col_count=(4, "fixed"),
              interactive=False,
              elem_id="danh_sach_sach_table"
          )
          dich(danh_sach_sach, value=lambda l: _sang_dataframe(doc_du_lieu_sach(), l))

      gr.Markdown("---")

      # -------------- CẬP NHẬT TIẾN ĐỘ / TRẠNG THÁI --------------
      with dich(gr.Accordion("✏️ Cập nhật tiến độ đọc", open=False), label="lt_cap_nhat_td"):
          chon_sach_cap_nhat = gr.Dropdown(
              label="Chọn sách",
              choices=[],
              interactive=True
          )

          with gr.Row():
              trang_hien_tai_cap_nhat = gr.Number(
                  label="Trang hiện tại",
                  value=0,
                  minimum=0,
                  scale=1
              )
              trang_thai_cap_nhat = gr.Dropdown(
                  label="Trạng thái",
                  choices=["Chưa đọc", "Đang đọc", "Đã đọc xong"],
                  scale=1
              )
              yeu_thich_cap_nhat = gr.Checkbox(
                  label="⭐ Yêu thích",
                  scale=0
              )

          with gr.Row():
              gr.HTML("")  # spacer
              cap_nhat_btn = gr.Button("🔄 Cập nhật", variant="primary", scale=0, min_width=150)
              xoa_sach_btn = gr.Button("🗑️ Xoá sách", variant="stop", scale=0, min_width=150)
          dich(chon_sach_cap_nhat, label="lt_chon_sach")
          dich(trang_hien_tai_cap_nhat, label="lt_trang")
          dich(trang_thai_cap_nhat, label="lt_trang_thai", choices="lt_ds_trang_thai")
          dich(yeu_thich_cap_nhat, label="lt_yt")
          dich(cap_nhat_btn, value="lt_cap_nhat")
          dich(xoa_sach_btn, value="lt_xoa")

######################### CÀI ĐẶT #########################
    with gr.Group(visible=False, elem_id="trang_cai_dat") as Cai_Dat:
      # gr.Markdown("## Cài đặt sớm sẽ được cập nhật")    # (bản cũ) -> nay đã có các mục cài đặt bên dưới
      dich(gr.HTML(render_cd_banner()), value=render_cd_banner)

      with gr.Row(equal_height=True):
          with gr.Column(elem_classes="cd-the"):
              dich(gr.Markdown(_chu("cd_ngon_ngu")), value="cd_ngon_ngu")
              chon_ngon_ngu = gr.Radio(choices=[("Tiếng Việt", "vi"), ("English", "en")], value="vi",
                                       show_label=False, container=False, elem_classes="cd-chon")
          with gr.Column(elem_classes="cd-the"):
              dich(gr.Markdown(_chu("cd_giao_dien")), value="cd_giao_dien")
              chon_giao_dien = dich(gr.Radio(choices=list(zip(_chu("cd_ds_giao_dien"), TU_DIEN["cd_ds_giao_dien"]["gia_tri"])),
                                             value="sang", show_label=False, container=False, elem_classes="cd-chon"),
                                    choices="cd_ds_giao_dien")

      with gr.Row(equal_height=True):
          with gr.Column(elem_classes="cd-the"):
              dich(gr.Markdown(_chu("cd_co_chu")), value="cd_co_chu")
              chon_co_chu = dich(gr.Radio(choices=list(zip(_chu("cd_ds_co_chu"), TU_DIEN["cd_ds_co_chu"]["gia_tri"])),
                                          value="vua", show_label=False, container=False, elem_classes="cd-chon"),
                                 choices="cd_ds_co_chu")
          with gr.Column(elem_classes="cd-the"):
              dich(gr.Markdown(_chu("cd_hieu_ung")), value="cd_hieu_ung")
              bat_hieu_ung = dich(gr.Checkbox(value=True, label=_chu("cd_bat_hieu_ung")), label="cd_bat_hieu_ung")

      with gr.Row(equal_height=True):
          with gr.Column(scale=3):
              dich(gr.HTML(render_cd_thong_tin()), value=render_cd_thong_tin)
              dich(gr.Markdown(_chu("cd_luu_y")), value="cd_luu_y")
          khoi_phuc_btn = dich(gr.Button("↺ Khôi phục mặc định", variant="secondary", scale=1), value="cd_khoi_phuc")




################################################################################

# CÁC SỰ KIỆN

################################################################################

    ####################  TỔNG THỂ   ####################

    # -------------- CHUYỂN TRANG  --------------
    #Trang chủ
    trang_chu.click(
    fn=lambda: set_active_page("trang_chu"),
    inputs=[],
    outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn]
    )

    #ChatBox
    btn_AI.click(
    fn=lambda: set_active_page("chatbox"),
    inputs=[],
    outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn]
    ).then(
    fn=lam_moi_sidebar,
    inputs=[luu_tru_chat, chat_id_hien_tai],
    outputs=[danh_sach_lich_su]
    )

    #Lưu Trữ & load danh sách (giống bản MQ_v2: mở trang là thấy bảng sách ngay)
    luu_tru.click(
    fn=lambda: set_active_page("luu_tru"),
    outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn]
    ).then(
        fn=hien_thi_tat_ca_sach,
        inputs=[chon_ngon_ngu],
        outputs=[danh_sach_sach, chon_sach_cap_nhat]
    )
    cai_dat.click(
    fn=lambda: set_active_page("cai_dat"),
    outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn]
    )

    # -------------- THIẾT LẬP --------------
    # Thiết lập trạng thái ban đầu khi ứng dụng tải lên
    Library.load(
        fn=lambda: set_active_page("trang_chu"),
        inputs=[],
        outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn],
        queue=False
    )
    # -------------- CÀI ĐẶT: nạp lại lựa chọn đã lưu trên trình duyệt --------------
    CAI_DAT_VAO = [chon_ngon_ngu, chon_giao_dien, chon_co_chu, bat_hieu_ung]
    DICH_RA = [comp for comp, _ in DS_DICH]

    Library.load(
        fn=nap_cai_dat, inputs=[cai_dat_luu], outputs=CAI_DAT_VAO
    ).then(
        fn=None, inputs=CAI_DAT_VAO, js=JS_AP_DUNG_CAI_DAT          # sáng/tối, cỡ chữ, hiệu ứng
    ).then(
        fn=nap_ngon_ngu, inputs=[chon_ngon_ngu], outputs=DICH_RA
    ).then(
        fn=doi_nhom_dna, inputs=[dna_nhom, dna_chon, chon_ngon_ngu], outputs=[dna_chon]
    )

    #################### TRANG CHỦ  ####################


    # -------------- Sự kiện tìm kiếm --------------
    inputs_search = [search, chu_de_box, sap_xep, chon_ngon_ngu]

    # Thanh tìm kiếm
    search.change(ap_dung, inputs_search,  [goi_y_box, output_trang_chu, popup_loc])

    sap_xep.change(ap_dung, inputs_search,  [goi_y_box, output_trang_chu, popup_loc])

    # Bộ lọc
    loc_btn.click(lambda: gr.update(visible=True), None, popup_loc)

    dong_loc_btn.click(lambda: gr.update(visible=False), None, popup_loc)

    xoa_btn.click(lambda: [], None, chu_de_box)

    tim_btn.click(ap_dung,
                  [search, chu_de_box, sap_xep, chon_ngon_ngu],
                  [goi_y_box, output_trang_chu, popup_loc])

    # Ảnh
    anh_btn.click(lambda: gr.update(visible=True), None, popup_anh)

    dong_anh_btn.click(lambda: gr.update(visible=False), None, popup_anh)

    # --- tải ảnh: lưu ảnh gốc, xoá vùng cũ ---
    img_input.upload(khi_tai_anh, img_input, [anh_goc, diem_chon, vung_status])

    # --- bấm lên ảnh để chọn 2 góc ---
    img_input.select(khi_bam, [anh_goc, diem_chon], [img_input, diem_chon, vung_status])

    # Xóa vùng
    xoa_vung_btn.click(xoa_vung, anh_goc, [img_input, diem_chon, vung_status])

    # --- đọc chữ: GỘP HẾT các ô text -> thanh tìm kiếm ---
    ocr_btn.click(bam_doc, [anh_goc, diem_chon, search, chu_de_box, sap_xep, chon_ngon_ngu],
                 [search, output_trang_chu, vung_status])


    # --- DNA tác giả ---
    dna_btn.click(
        fn=lambda: set_active_page("dna"),
        inputs=[],
        outputs=[Trang_Chu, Chatbox, Luu_Tru, Cai_Dat, Trang_DNA,
                 trang_chu, btn_AI, luu_tru, cai_dat, dna_btn]
    ).then(fn=render_dna, inputs=[dna_chon, chon_ngon_ngu], outputs=dna_html)

    dna_chon.change(fn=render_dna, inputs=[dna_chon, chon_ngon_ngu], outputs=dna_html)

    # Lọc nhóm tác giả -> đổi danh sách trong Dropdown
    dna_nhom.input(fn=doi_nhom_dna, inputs=[dna_nhom, dna_chon, chon_ngon_ngu], outputs=dna_chon)

    # ◀ / 🎲 / ▶
    dna_truoc.click(fn=lambda ht, nh: dna_ke_ben(ht, nh, -1), inputs=[dna_chon, dna_nhom], outputs=dna_chon)
    dna_sau.click(fn=lambda ht, nh: dna_ke_ben(ht, nh, 1), inputs=[dna_chon, dna_nhom], outputs=dna_chon)
    dna_ngau.click(fn=dna_ngau_nhien, inputs=[dna_chon, dna_nhom], outputs=dna_chon)

    # Bấm "tác giả cùng gen" trong HTML (JS ghi vào ô ẩn dna_cau)
    dna_cau.change(fn=dna_tu_cau_noi, inputs=[dna_cau, chon_ngon_ngu], outputs=[dna_nhom, dna_chon])

    ####################  ChatBot ####################

    # Sự kiện khi gửi tin nhắn (Bổ sung thêm inputs và outputs)
    nut_gui.click(
        fn=xu_ly,
        inputs=[o_nhap, chat, luu_tru_chat, chat_id_hien_tai, chon_ngon_ngu],
        outputs=[chat, o_nhap, luu_tru_chat, chat_id_hien_tai, danh_sach_lich_su]
    )
    o_nhap.submit(
        fn=xu_ly,
        inputs=[o_nhap, chat, luu_tru_chat, chat_id_hien_tai, chon_ngon_ngu],
        outputs=[chat, o_nhap, luu_tru_chat, chat_id_hien_tai, danh_sach_lich_su]
    )

    # Sự kiện khi bấm nút "+ Sổ ghi chú mới"
    btn_them_moi.click(
        fn=tao_chat_moi,
        inputs=None,
        outputs=[chat, o_nhap, chat_id_hien_tai, danh_sach_lich_su]
    )

    # Sự kiện khi bấm vào 1 dòng trong danh sách lịch sử
    danh_sach_lich_su.select(
        fn=mo_lich_su_cu,
        inputs=[danh_sach_lich_su, luu_tru_chat],
        outputs=[chat, chat_id_hien_tai]
    )

    # (Tính năng Like giữ nguyên)
    chat.like(print_like_dislike, None, None)
    ####################  LƯU TRỮ  ####################

    # Thêm sách mới
    luu_sach_btn.click(
        fn=luu_sach_moi,
        inputs=[ten_sach_moi, trang_hien_tai_moi, trang_thai_moi, yeu_thich_moi, chon_ngon_ngu],
        outputs=[danh_sach_sach, chon_sach_cap_nhat,
                 ten_sach_moi, trang_hien_tai_moi, trang_thai_moi, yeu_thich_moi]
    )

    # Bộ lọc: trạng thái + yêu thích (kết hợp)
    chua_doc_btn.click(
        fn=lambda yt, nn: loc_sach("Chưa đọc", yt, nn),
        inputs=[yeu_thich_checkbox, chon_ngon_ngu],
        outputs=[danh_sach_sach]
    )
    dang_doc_btn.click(
        fn=lambda yt, nn: loc_sach("Đang đọc", yt, nn),
        inputs=[yeu_thich_checkbox, chon_ngon_ngu],
        outputs=[danh_sach_sach]
    )
    da_xong_btn.click(
        fn=lambda yt, nn: loc_sach("Đã đọc xong", yt, nn),
        inputs=[yeu_thich_checkbox, chon_ngon_ngu],
        outputs=[danh_sach_sach]
    )
    tat_ca_btn.click(
        fn=lambda yt, nn: loc_sach(None, yt, nn),
        inputs=[yeu_thich_checkbox, chon_ngon_ngu],
        outputs=[danh_sach_sach]
    )

    # (Tùy chọn) Khi người dùng tick/bỏ tick checkbox, tự động lọc lại ngay
    # — cần biết đang xem trạng thái nào, nên đơn giản nhất là để checkbox
    #   tự áp dụng lên toàn bộ danh sách hiện có (không giữ trạng thái đã lọc trước đó):
    yeu_thich_checkbox.change(
        fn=lambda yt, nn: loc_sach(None, yt, nn),
        inputs=[yeu_thich_checkbox, chon_ngon_ngu],
        outputs=[danh_sach_sach]
    )

    # Khi chọn 1 sách trong Dropdown cập nhật -> tự động điền thông tin hiện có
    chon_sach_cap_nhat.change(
        fn=tai_thong_tin_sach,
        inputs=[chon_sach_cap_nhat],
        outputs=[trang_hien_tai_cap_nhat, trang_thai_cap_nhat, yeu_thich_cap_nhat]
    )

    # Cập nhật tiến độ đọc
    cap_nhat_btn.click(
        fn=cap_nhat_tien_do,
        inputs=[chon_sach_cap_nhat, trang_hien_tai_cap_nhat, trang_thai_cap_nhat, yeu_thich_cap_nhat, chon_ngon_ngu],
        outputs=[danh_sach_sach, chon_sach_cap_nhat]
    )

    # Xoá sách
    xoa_sach_btn.click(
        fn=xoa_sach_khoi_kho,
        inputs=[chon_sach_cap_nhat, chon_ngon_ngu],
        outputs=[danh_sach_sach, chon_sach_cap_nhat,
                 trang_hien_tai_cap_nhat, trang_thai_cap_nhat, yeu_thich_cap_nhat]
    )

    ####################  CÀI ĐẶT  ####################

    # Sáng/tối, cỡ chữ, hiệu ứng: JS áp dụng ngay trên trình duyệt rồi Python lưu lại
    for comp in (chon_giao_dien, chon_co_chu, bat_hieu_ung):
        comp.input(fn=luu_cai_dat, inputs=CAI_DAT_VAO, outputs=cai_dat_luu, js=JS_AP_DUNG_CAI_DAT)

    # Đổi ngôn ngữ: lưu -> dịch toàn bộ giao diện -> vẽ lại DNA & kết quả tìm kiếm theo ngôn ngữ mới
    chon_ngon_ngu.input(
        fn=luu_cai_dat, inputs=CAI_DAT_VAO, outputs=cai_dat_luu
    ).then(
        fn=doi_ngon_ngu, inputs=[chon_ngon_ngu], outputs=DICH_RA
    ).then(
        fn=doi_nhom_dna, inputs=[dna_nhom, dna_chon, chon_ngon_ngu], outputs=[dna_chon]
    ).then(
        fn=render_dna, inputs=[dna_chon, chon_ngon_ngu], outputs=dna_html
    ).then(
        fn=ap_dung, inputs=inputs_search, outputs=[goi_y_box, output_trang_chu, popup_loc]
    )

    # Khôi phục mặc định
    khoi_phuc_btn.click(
        fn=khoi_phuc_cai_dat, inputs=None, outputs=CAI_DAT_VAO
    ).then(
        fn=luu_cai_dat, inputs=CAI_DAT_VAO, outputs=cai_dat_luu, js=JS_AP_DUNG_CAI_DAT
    ).then(
        fn=doi_ngon_ngu, inputs=[chon_ngon_ngu], outputs=DICH_RA
    ).then(
        fn=doi_nhom_dna, inputs=[dna_nhom, dna_chon, chon_ngon_ngu], outputs=[dna_chon]
    ).then(
        fn=render_dna, inputs=[dna_chon, chon_ngon_ngu], outputs=dna_html
    )


if __name__ == "__main__":
    Library.launch(share=True, debug=True, show_error=True)
```
